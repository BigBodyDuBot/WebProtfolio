### Example 1
 Find names of sailors who’ve reserved boat #103

```python
%%sql
SELECT S.sname
FROM Sailors S
JOIN Reserves R ON S.sid = R.sid
WHERE R.bid=103;
```

### Example 2
Find names of sailors whose rating are larger than 7 and have reserved a red boat. List each matching sailor name once.

```python
%%sql
SELECT DISTINCT S.sname
FROM Sailors S
JOIN Reserves R ON S.sid = R.sid
JOIN Boats B ON R.bid = B.bid
WHERE S.rating > 7 AND B.color = 'red';
```

### Exercise 1

Find all products under $200 manufactured in Japan, and return their names and prices.

product(<u>pname</u> : varchar, Price: real, category: varchar, manufacturer:varchar)

company(<u>cname</u>: varchar, stock: integer, country: varchar)

```python
%%sql
SELECT P.pname, P.price
FROM product P, company C
WHERE P.manufacturer = C.cname and P.price <200;
```

### Exercise 2
Find sailor ids who’ve reserved a red or a green boat; list each matching sailor id once.

sailors(<u>sid</u>:integer, sname:string, rating:integer, age:real)

reserves(<u>sid</u>:integer, <u>bid</u>:integer, day:date)

boats(<u>bid</u>:integer, bname:string, color:string)

```python
%%sql
SELECT DISTINCT r.sid
FROM reserves r
JOIN boats b ON r.bid = b.bid
WHERE b.color = 'red' OR b.color = 'green';
```

### Exercise 3
Find sailor ids who’ve reserved a red and a green boat

sailors(<u>sid</u>:integer, sname:string, rating:integer, age:real)

reserves(<u>sid</u>:integer, <u>bid</u>:integer, day:date)

boats(<u>bid</u>:integer, bname:string, color:string)

```python
%%sql
SELECT DISTINCT r1.sid
FROM reserves r1
JOIN boats b1 ON r1.bid = b1.bid
JOIN reserves r2 ON r1.sid = r2.sid
JOIN boats B2 ON r2.bid = b2.bid
WHERE b1.color ='red' AND b2.color ='green';
```
