## **Q1 (4 points)**

Find all customers who are located in New York and display all of their information.

```python
%%sql
SELECT *
FROM Customers
WHERE city='New York';
```

## **Q2 (4 points)**

Retrieve only the product name and price of every product in the catalog.

```python
%%sql
SELECT product_name, price
FROM Products;
```

## **Q3 (4 points)**

Generate all combinations of customer C001's profile with products P101 and P102. (Hint: use cross product)

```python
%%sql
SELECT *
FROM
    (SELECT * FROM Customers WHERE cid = 'C001') ,
    (SELECT * FROM Products WHERE pid IN ('P101', 'P102')) ;
```

## **Q4 (4 points)**

Find all orders where the quantity is greater than 1, and include the product name and price for each order.

```python
%%sql

SELECT p.product_name, p.price
FROM Orders o
JOIN Products p ON o.pid = p.pid
WHERE o.quantity > 1;
```

## **Q5 (4 points)**

Find all orders for products in the Electronics category. Display the order id, customer id, product name, and order date.

```python
%%sql

SELECT o.oid, o.cid, p.product_name, o.order_date
FROM Orders o
JOIN Products p ON o.pid = p.pid
WHERE p.category = 'Electronics';

```

## **Q6 (4 points)**

Join the Customers table and the Orders table and display all columns.

```python
%%sql

SELECT *
FROM Customers c
JOIN Orders o
ON c.cid = o.cid;
```

## **Q7 (4 points)**

Retrieve all products with a price greater than $100 and rename the output columns as: product_id, name, cat, cost.

```python
%%sql

SELECT
    pid AS product_id,
    product_name AS name,
    category AS cat,
    price AS cost
FROM Products
WHERE price > 100;
```

## **Q8 (4 points)**

Find the name and city of all customers whose name starts with the letter "A" or ends with "h". Display the results with column headers customer_name and customer_city.

```python
%%sql

SELECT
    name AS customer_name,
    city AS customer_city
FROM Customers
WHERE name LIKE 'A%'
   OR name LIKE '%h';
```

## **Q9 (4 points)**

For each order, display the customer name, product name, category, and order date. Only show orders where the product price is above $100.

```python
%%sql

SELECT
    c.name,
    p.product_name,
    p.category,
    o.order_date
FROM Orders o
JOIN Customers c ON o.cid = c.cid
JOIN Products p ON o.pid = p.pid
WHERE p.price > 100;
```

## **Q10 (4 points)**

Find the distinct cities of Prime members who have placed an order for an Electronics product.

```python
%%sql

SELECT DISTINCT c.city
FROM Customers c
JOIN Orders o  ON c.cid = o.cid
JOIN Products p ON o.pid = p.pid
WHERE c.membership = 'Prime'
  AND p.category = 'Electronics';
```
